# Preprocessing dataset

This notebook contains steps for processing available dataset to find insights and make it suitable for training

---

## Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

## Import dataset

```python
df = pd.read_csv('dataset.csv')
df.head()
```

```python
df.info()
```

The dataset contains values that need transformation before training. 

## Preprocessing

---

- `model_year`

```python
df.model_year.unique()
```

The 'model_year' column is in thousands, converting it in 'age' that can reduce the scale

```python
df['vehicle_age'] = 2026 - df.model_year
df.head()
```

```python
df = df.drop(columns=['model_year'])
df.head()
```

---

- `milage`

```python
df.milage.unique()
```

Converting milage values into integer after removing comma and string

```python
df['milage'] = df.milage.str.replace(r"[^0-9]", "", regex=True).astype(int)
```

```python
df.head()
```

Scale milage column to a smaller value

```python
scaler = StandardScaler()
df['milage'] = scaler.fit_transform(df[['milage']])
```

```python
df.head()
```

---

- `fuel_type`

```python
df['fuel_type'].unique()
```

There are 'nan' values

```python
df.fillna("Unknown", inplace=True)
df[df['fuel_type'] == "Unknown"]
```

'nan' values represent electric vehicles. Columns containing these values cannot be dropped

Label encoding `fuel_type` values

```python
le = LabelEncoder()
df['fuel_type'] = le.fit_transform(df['fuel_type'])
df.head()
```

```python
df['fuel_type'].unique()
```

---

- `engine`

```python
df['engine'].unique()
```

The `engine` column has many unique values. It has useful insights including HP, engine cylinder and liters. Extract these important features

Extract HP, engine size, and cylinders into separate columns

```python
df['HP'] = df['engine'].str.extract(r'(\d+\.?\d*)HP').astype(float)
```

```python
df['engine_size'] = df['engine'].str.extract(r'(\d+\.?\d*)L').astype(float)
```

```python
df['cylinders'] = df['engine'].str.extract(r'V(\d+)').astype(float)
```

```python
df.head()
```

```python
df['cylinders'].isnull().sum()
```

`cylinders` has lot of null values, so drop the column

```python
df = df.drop(columns=['cylinders'])
```

```python
df['HP'].isnull().sum()
```
